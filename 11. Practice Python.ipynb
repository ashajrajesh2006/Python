```python
#to generate the square of every element of a list.
numbers = [1, 2, 3, 4]
squares = [x**2 for x in numbers]
print(squares)
```

```python
#to simplify code that generate a list of square for all even number from 0 to 10?
squares = [x**2 for x in range(11) if x % 2 == 0]
print(squares)
```

```python
#Convert string to list
string = "Asha"
string_list = list(string)
print(string_list)
```

```python
#negative indexing in python-Negative indexing allows access to elements from the end of a list.
my_list = [1, 2, 3, 4,5,6,7]
print(my_list[-1])
```

```python
#append() and extend()
l = [1, 2, 3,4,5]
l.append([6, 7])
print(l)
```

```python
l = [1, 2, 3,4,5]
l.extend([6, 7])
print(l)
```
