```python
age=int(input("Enter the age: "))
if(age>=18):
    print("Adult")
else:
    print("Minor")
```

```python
# If-elif
age=int(input("Enter the age: "))
if(age<18):
    print("Minor")
elif(age<60):
    print("Adult")
else:
    print("Senior Citizen")

    
```

```python
# For loop 
lists=[25,14,67]
for age in lists:
    if(age<18):
            print("Minor")
    elif(age<60):
            print("Adult")
else:
        print("Senior Citizen")

```
