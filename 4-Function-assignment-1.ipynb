```python
# Create a function that checks whether the given number is Odd or Even
def oddeven():
    num=int(input("Enter the number: "))
    if((num%2)==1):
        print("Odd Number")
        message=("Odd Number")
    else:
        print("even number")
        message=("even number")
    return message
```
